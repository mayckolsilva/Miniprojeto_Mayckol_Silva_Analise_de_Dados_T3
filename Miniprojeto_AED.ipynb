```python
# importar as bibliotecas necessárias para o projeto

import pandas as pd
import numpy as np
```

```python
# Carregar as informações do arquivo Varejo.csv para apresentação dos dados
df_varejo = pd.read_csv("Base Varejo.csv", encoding="UTF-8", sep=";")

# Remoção de colunas vazias e que não sejam apresentadas
df_varejo = df_varejo.dropna(axis=1, how="all")

base = len(df_varejo)
# Números de Registros
print("Números de Registros:", base, "\n")
print("-"*30)

# Apresentar Colunas
print("Apresentação de Colunas: \n", df_varejo.columns, "\n")
print("-"*30)

# Apresentar tipos de Dados
print (df_varejo.dtypes)
print("-"*30)
```

### Reportar possíveis erros

```python
# Identificação de valores nulos e contagem
print(df_varejo.isnull())
print(df_varejo.isnull().sum())
```

```python
# Identificação de linhas duplicadas e contagem
print(df_varejo.duplicated())
print("Há um total de ", df_varejo.duplicated().sum(), " linhas duplicadas")

# Remoção de linhas duplicadas

df_varejo = df_varejo.drop_duplicates()
print(f"Duplicadas removidas: {base - len(df_varejo)}")
```

```python
# Verificação da data das vendas formato correto.
df_varejo["DATA"] = pd.to_datetime(df_varejo["DATA"], errors="coerce", dayfirst=True)
qnt_datas_vazias = df_varejo["DATA"].isnull().sum()
print(f"Quantidade de Datas Vazias: {qnt_datas_vazias}")

# Remoção de Vendas onde não possuem data
df_varejo = df_varejo.dropna(subset=["DATA"])
pos_remocao = len(df_varejo)


print(f"Linhas removidas por data vazia: {base - pos_remocao}")
print(f"Total de Registros após remoção : {pos_remocao}")
```

```python
# Verificação de valores únicos para a coluna Categoria e então identificar informações
print(df_varejo["PR_CAT"].unique())
```

```python
# Identificação de categorias nulas = sem qualquer caractere
categoria_nulos = df_varejo["PR_CAT"].isnull().sum()
print(f"Quantidade de linhas nulas na Coluna Categoria: {categoria_nulos}")
```

```python
# Caso a categoria esteja com espaço vazio fará a identificação também e então somar

categoria_espaco_vazio = df_varejo["PR_CAT"].astype(str).str.strip().eq("").sum()
print(f"Quantidade de linhas vazias (com espaço) na Coluna Categoria: {categoria_espaco_vazio}")

# para dados do tipo #NaN, transformei em "nan" string.
# Espaços no início e fim eu removi
# Precisei equiparar à espaço vazio com ""
# E por fim, somei.
```

```python
# Identificar linhas que estão como #N/D e também fazer a contagem.
categoria_nd = df_varejo["PR_CAT"].astype(str).str.strip().eq("#N/D").sum()
print(f"Quantidade de linhas com  '#N/D' na Coluna Categoria: {categoria_nd}")

# Mesmo processo só que para "#N/D"

# Agora ajusto de #N/D para "Não Informado".
df_varejo["PR_CAT"] = df_varejo["PR_CAT"].replace("#N/D", "Não Informado")
```

```python
# Produtos devem conter todas as letras maiúscula, sem espaços na frente e final
df_varejo["PR_NOME"] = df_varejo["PR_NOME"].str.strip()
df_varejo["PR_NOME"] = df_varejo["PR_NOME"].str.upper()
```

```python
# Identificação da coluna de número de filhos agrupado pelo ID do cliente.
qnt_filhos_cliente = df_varejo.groupby("CL_ID")["CL_FHL"].first()
```

```python
# Estatística referente à quantidade de filhos por cliente.
print("-"*30)
print (f"Contagem de Filhos Por Cliente: {qnt_filhos_cliente.count()}")
print("-"*30)

print (f"Média de Filhos Por Cliente: {qnt_filhos_cliente.mean():.2f}")
print("-"*30)

print (f"Mediana de Filhos Por Cliente: {qnt_filhos_cliente.median():.2f}")
print("-"*30)

print (f"Desvio Padrão de Filhos Por Cliente: {qnt_filhos_cliente.std():.2f}")
print("-"*30)

print (f"Moda de Filhos Por Cliente: {qnt_filhos_cliente.mode()[0]}")
print("-"*30)

print (f"Mínimo de Filhos Por Cliente: {qnt_filhos_cliente.min()}")
print("-"*30)

print (f"Máximo de Filhos Por Cliente: {qnt_filhos_cliente.max()}")
print("-"*30)

print (f"1º Quartil (25%): {qnt_filhos_cliente.quantile(0.25)}")
print("-"*30)

print (f"3º Quartil (75%): {qnt_filhos_cliente.quantile(0.75)}")
print("-"*30)
```

```python
# Identificação de Vendas por Gênero

vnd_por_genero = df_varejo.groupby("CL_GENERO")["CO_ID"].count()
print (vnd_por_genero)

#Vendas Masculinos
print("-"*30)
print (f"Vendas Masculino: {vnd_por_genero.get("M",0)}")
print("-"*30)

print (f"Vendas Feminino: {vnd_por_genero.get("F",0)}")
print("-"*30)
```

```python
# Identificação das vendas por categoria
vnd_por_categoria = df_varejo.groupby("PR_CAT")["CO_ID"].count()
print(f"Quantidade de Vendas por Categoria: \n {vnd_por_categoria}")
print("-"*30)

# Identificação do percentual das vendas por categoria
percent_categoria = (vnd_por_categoria / vnd_por_categoria.sum()*100).round(2)
print(f"Percentual de Vendas por Categoria:\n {percent_categoria}")
```

## Conclusões

- **Qualidade dos dados**: a base original continha 830.000 registros, dos quais 96.553 (11,6%) eram duplicatas e foram removidas. Após a limpeza (duplicatas e datas inválidas), restaram 733.447 registros válidos para análise.

- **Categorias de produto**: 3.228 registros (0,44%) estavam marcados como "#N/D" na coluna de categoria e foram padronizados como "Não Informado", preservando o registro em vez de descartá-lo.

- **Categorias mais vendidas**: ALIMENTOS domina as vendas (384.197 registros, 52,38% do total), seguido por HIGIENE (137.702) e LIMPEZA (128.632). ACESSORIOS é a categoria com menor volume (12.871), sugere que há um mix de produtos e é fortemente concentrado em itens de consumo básico.

- **Perfil de filhos dos clientes**: a média é de 1,14 filhos por cliente, mas a mediana é 0, isso mostra que a maioria dos clientes não tem filhos, e a média é ajustada para cima por um grupo menor com até 4 filhos.

- **Vendas por gênero**: clientes do gênero feminino representam a maioria das vendas (382.427 vs. 351.020 do masculino), uma diferença de cerca de 8%.

- **Limitações e problemas remanescentes**: Mesmo que tenha feito limpeza, a base ainda pode conter inconsistências não verificadas, como valores de "CL_FHL" ou "CL_EC" fora de faixas plausíveis.
